# Sinhala Sign Language (SSL) - Comprehensive Dataset Analysis

This notebook provides a complete breakdown of the Sinhala Sign Language (SSL) dataset, feature representation, augmentation pipeline, and semantic categories.

### Dataset Highlights:
- **Total Gesture Classes:** 171
- **Total Processed Sequences:** 2,052 sequences (171 base recorded videos × 12 = 1 original + 11 augmentations)
- **Sequence Window:** 30 frames (1.0 second at 30 FPS)
- **Feature Vector Dimension:** 1,629 features per frame

### MediaPipe Holistic Spatial Decomposition:
1. **Face Mesh:** 468 landmarks × (x, y, z) = **1,404 features**
2. **Pose Tracking:** 33 landmarks × (x, y, z) = **99 features**
3. **Left Hand:** 21 landmarks × (x, y, z) = **63 features**
4. **Right Hand:** 21 landmarks × (x, y, z) = **63 features**
$$\text{Total Features per Frame} = 1404 + 99 + 63 + 63 = \mathbf{1629}$$
$$\text{Input Shape per Sample} = (30, 1629)$$

```python
import json
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data_dir = Path("../data/processed")
raw_dir = Path("../data/raw")

# Load processed metadata and labels
with open(data_dir / "metadata.json", "r", encoding="utf-8") as f:
    metadata = json.load(f)

labels = np.load(data_dir / "labels.npy", allow_pickle=True)
X = np.load(data_dir / "processed_data.npy", mmap_mode='r')

print(f"Dataset Shape:               {X.shape}")
print(f"Total Processed Sequences:   {len(labels)}")
print(f"Unique Gesture Classes:      {len(set(labels))}")
print(f"Augmentations per video:     {metadata.get('augmentations_per_video')}")
```

## 7 Semantic Categories Breakdown
Inspection of the raw dataset taxonomy in `data/raw`.

```python
categories = {}
for category_dir in raw_dir.iterdir():
    if category_dir.is_dir():
        vids = list(category_dir.glob("*.mp4"))
        categories[category_dir.name] = len(vids)

cat_df = pd.DataFrame(list(categories.items()), columns=["Category", "Raw Videos"])
cat_df["Total Augmented (12x)"] = cat_df["Raw Videos"] * 12
display(cat_df)

# Bar chart
plt.figure(figsize=(10, 4.5))
plt.bar(cat_df["Category"], cat_df["Raw Videos"], color="#3b82f6")
plt.title("Class Distribution Across Semantic Categories", fontweight="bold")
plt.ylabel("Number of Unique Signs")
plt.xticks(rotation=30, ha="right")
plt.grid(True, linestyle="--", alpha=0.5, axis="y")
plt.tight_layout()
plt.show()
```
