# Sinhala Sign Language (SSL) - Dataset Exploration

This notebook inspects the raw video corpus and the preprocessed 1629-dimensional MediaPipe landmark dataset used for training the real-time gesture recognition model.

```python
import json
import numpy as np
from pathlib import Path
import matplotlib.pyplot as plt

data_dir = Path("../data/processed")
meta_path = data_dir / "metadata.json"
labels_path = data_dir / "labels.npy"

with open(meta_path, "r", encoding="utf-8") as f:
    metadata = json.load(f)

labels = np.load(labels_path, allow_pickle=True)
print(f"Total Samples: {metadata.get('num_samples')}")
print(f"Unique Gesture Classes: {len(set(labels))}")
print(f"Augmentations per class: {metadata.get('augmentations_per_video')}")
```

## Class Distribution
Analyzing sample distribution across categories (numbers, alphabet, verbs, sentences).

```python
unique_classes, counts = np.unique(labels, return_counts=True)
print(f"Average samples per class: {counts.mean():.1f}")

# Display first 15 classes
for u, c in zip(unique_classes[:15], counts[:15]):
    print(f"- {u}: {c} samples")
```
