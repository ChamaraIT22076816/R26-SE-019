# Model Training Workflow: LSTM, Bi-LSTM & Transformer Architectures

This notebook documents the training workflow, architectural designs, and parameter counts for the 3 temporal sequence models implemented for Sinhala Sign Language recognition:

1. **Stacked LSTM** (`LSTMRecognizer`)
2. **Bidirectional LSTM** (`build_bidirectional_lstm`)
3. **Spatial-Temporal Transformer** (`SignTransformerRecognizer`)

```python
import yaml
import sys
from pathlib import Path

ROOT = Path("..").resolve()
sys.path.insert(0, str(ROOT))
sys.path.insert(0, str(ROOT / "src"))

config_path = ROOT / "config" / "config.yaml"
with open(config_path, "r", encoding="utf-8") as f:
    config = yaml.safe_load(f)

config['dataset']['num_classes'] = 171

from model.lstm_model import LSTMRecognizer
from model.transformer_model import SignTransformerRecognizer

lstm_builder = LSTMRecognizer(config)
lstm_model = lstm_builder.build_model()
bilstm_model = lstm_builder.build_bidirectional_lstm()

trans_builder = SignTransformerRecognizer(config)
trans_model = trans_builder.build_model()

print(f"1. Stacked LSTM Parameters:       {lstm_model.count_params():,}")
print(f"2. Bidirectional LSTM Parameters: {bilstm_model.count_params():,}")
print(f"3. Transformer Parameters:        {trans_model.count_params():,}")
```
