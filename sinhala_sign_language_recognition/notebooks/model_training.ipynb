# Model Training Workflow: LSTM & Transformer Architectures

This notebook documents the training workflow for temporal sequence models on SSL landmark sequences.

```python
import yaml
from pathlib import Path

config_path = Path("../config/config.yaml")
with open(config_path, "r") as f:
    config = yaml.safe_load(f)

print("Model Type:", config['model']['type'])
print("Sequence Length:", config['model']['sequence_length'])
print("Num Features:", config['model']['num_features'])
print("Epochs:", config['training']['epochs'])
print("Batch Size:", config['training']['batch_size'])
print("Classes:", config['dataset']['num_classes'])
```

## Architectures Implemented
1. **Stacked LSTM** (`LSTMRecognizer`): 256 -> 128 -> 64 with recurrent dropout and BatchNormalization.
2. **Bidirectional LSTM**: Captures forward and reverse temporal context.
3. **Spatial-Temporal Transformer** (`SignTransformerRecognizer`): Multi-Head Self-Attention with sinusoidal positional encodings.
4. **FastInferenceEngine**: Multi-tier execution with TFLite and compiled `@tf.function`.
