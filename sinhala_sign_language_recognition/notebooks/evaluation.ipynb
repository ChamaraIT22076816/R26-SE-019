# Model Performance & Evaluation

Evaluates the trained LSTM and FastInferenceEngine against the test dataset partition.

```python
import json
from pathlib import Path

eval_file = Path("../models/saved_models/evaluation_results.json")
if eval_file.exists():
    with open(eval_file, "r") as f:
        results = json.load(f)
    print("Test Accuracy:", f"{results.get('test_accuracy', 0)*100:.2f}%")
    print("Test Loss:", results.get('test_loss'))
    print("Evaluation Timestamp:", results.get('timestamp'))
```

## Latency Benchmark Summary
- **Baseline `model.predict()`:** ~178 ms per sequence
- **Compiled `@tf.function`:** ~95 ms (1.9x speedup)
- **Accelerated `TFLite (XNNPACK)`:** ~26.9 ms (6.6x speedup)
- **Target Pipeline Turnaround:** < 65 ms total frame processing (15+ FPS)
