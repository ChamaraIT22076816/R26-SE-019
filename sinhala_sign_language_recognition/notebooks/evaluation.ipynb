# Multi-Model Performance & Evaluation
## Sinhala Sign Language (SSL) Recognition

This notebook evaluates the models trained on the SSL landmark corpus, comparing:
1. **Stacked LSTM** (Baseline)
2. **Bidirectional LSTM** (Bi-LSTM)
3. **Spatial-Temporal Transformer** (Multi-Head Self-Attention)

### Evaluation Metrics Tracked:
- Categorical Accuracy
- Precision (Macro & Weighted)
- Recall (Macro & Weighted)
- F1-Score (Macro & Weighted)
- Top-3 & Top-5 Categorical Accuracy
- Real-Time Inference Latency (ms per sequence)

```python
import json
import numpy as np
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt

saved_dir = Path("../models/saved_models")
comp_file = saved_dir / "model_comparison.json"

if comp_file.exists():
    with open(comp_file, "r", encoding="utf-8") as f:
        data = json.load(f)
    
    df = pd.DataFrame(data).T
    df['Accuracy (%)'] = (df['test_accuracy'] * 100).round(2)
    df['Macro F1 (%)'] = (df['f1_score_macro'] * 100).round(2)
    df['Weighted F1 (%)'] = (df['f1_score_weighted'] * 100).round(2)
    df['Top-3 Acc (%)'] = (df['top_3_accuracy'] * 100).round(2)
    df['Latency (ms)'] = df['inference_latency_ms'].round(1)
    df['Params'] = df['parameters'].apply(lambda x: f"{x:,}")
    
    cols = ['Params', 'Accuracy (%)', 'Macro F1 (%)', 'Weighted F1 (%)', 'Top-3 Acc (%)', 'Latency (ms)']
    display(df[cols])
else:
    print("Please run train_and_evaluate_all.py first to generate comparison metrics.")
```
